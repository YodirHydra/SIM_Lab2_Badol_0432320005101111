```python
import numpy as np
```

```python
# Student ID
student_id = "0432320005101111"

# Last digit of Student ID
d1 = int(student_id[-1])

# Second last digit of Student ID
d2 = int(student_id[-2])

print("Student ID:", student_id)
print("d1 =", d1)
print("d2 =", d2)
```

```python
A = np.array([
    [d1 + 2, d2 + 1],
    [2 * d1, d2 + 2]
])

print("Matrix A:")
print(A)
```

```python
print("Shape of A:", A.shape)
```

```python
det_A = np.linalg.det(A)

print("Determinant of A:", det_A)
```

```python
rank_A = np.linalg.matrix_rank(A)

print("Rank of A:", rank_A)
```

```python
eigenvalues_A = np.linalg.eigvals(A)

print("Eigenvalues of A:")
print(eigenvalues_A)
```

```python
if not np.isclose(det_A, 0):
    inverse_A = np.linalg.inv(A)
    print("Inverse of A:")
    print(inverse_A)
else:
    print("Matrix A is not invertible because determinant is zero.")
```

```python
B = A.copy()
B[0, 0] = B[0, 0] + 1

print("Changed element: A[0,0]")
print("Added +1 to A[0,0]")

print("\nMatrix B:")
print(B)
```

```python
print("Shape of B:", B.shape)
```

```python
det_B = np.linalg.det(B)

print("Determinant of B:", det_B)
```

```python
rank_B = np.linalg.matrix_rank(B)

print("Rank of B:", rank_B)
```

```python
eigenvalues_B = np.linalg.eigvals(B)

print("Eigenvalues of B:")
print(eigenvalues_B)
```

```python
if not np.isclose(det_B, 0):
    inverse_B = np.linalg.inv(B)
    print("Inverse of B:")
    print(inverse_B)
else:
    print("Matrix B is not invertible because determinant is zero.")
```

```python
print("----- Comparison of Matrix A and Matrix B -----")

print("Determinant of A:", det_A)
print("Determinant of B:", det_B)

print("Rank of A:", rank_A)
print("Rank of B:", rank_B)

print("Eigenvalues of A:", eigenvalues_A)
print("Eigenvalues of B:", eigenvalues_B)
```

1. How did the determinant change and why?

Answer:

The determinant of Matrix A was 5, while the determinant of Matrix B became 8. The determinant changed because one element of the matrix was increased by 1. Since the determinant depends on all matrix elements, changing one element changes the determinant.

2. Did the rank change? Explain.

Answer:

No, the rank did not change. The rank of both Matrix A and Matrix B is 2. Both matrices are full-rank 2×2 matrices because their determinants are non-zero.

3. How did the eigenvalues respond to the value change?

Answer:

The eigenvalues changed after increasing one element by 1. Matrix A had eigenvalues 5 and 1, while Matrix B had approximately 5.562 and 1.438. This shows that changing one matrix value can affect its eigenvalues.

4. Is B easier or harder to invert than A? Why?

Answer:

Both A and B are invertible because their determinants are non-zero. B is not fundamentally harder to invert than A because both are small 2×2 matrices. However, the determinant of B is larger, so B is also safely away from being singular.
